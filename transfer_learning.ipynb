```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
```

```python
IMG_SIZE=(224,224)
BATCH_SIZE=16
train_ds=tf.keras.utils.image_dataset_from_directory('dataset/train', image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='binary')
val_ds=tf.keras.utils.image_dataset_from_directory('dataset/validation', image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='binary')
print('Kelas:', train_ds.class_names)
```

```python
base=MobileNetV2(input_shape=IMG_SIZE+(3,), include_top=False, weights='imagenet')
base.trainable=False
model=models.Sequential([layers.Input(shape=IMG_SIZE+(3,)), layers.Rescaling(1./127.5, offset=-1), base, layers.GlobalAveragePooling2D(), layers.Dropout(0.2), layers.Dense(1,activation='sigmoid')])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history=model.fit(train_ds, validation_data=val_ds, epochs=10)
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['accuracy'], label='train')
plt.plot(history.history['val_accuracy'], label='validation')
plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.grid(); plt.show()
```

```python
model.save('models/nexus_mobilenetv2.keras')
print('Model tersimpan.')
```
